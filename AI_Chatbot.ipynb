```python
import json
import random
import ollama

from nltk.tokenize import word_tokenize
import nltk

nltk.download("punkt")
```

```python
with open("intents.json", "r") as file:
    data = json.load(file)

print(data["intents"])
```

```python
def get_intent_response(user_input):
    user_input = user_input.lower()

    for intent in data["intents"]:
        for pattern in intent["patterns"]:
            if pattern.lower() in user_input:
                return random.choice(intent["responses"])

    return None
```

```python
print(get_intent_response("hello"))
print(get_intent_response("thanks"))
print(get_intent_response("what time do you close"))
print(get_intent_response("what is artificial intelligence"))
```

```python
import ollama

print("=" * 60)
print("🤖 Hybrid AI Chatbot Started")
print("Type 'quit' to exit")
print("=" * 60)

messages = []

while True:

    user = input("You: ")

    if user.lower() == "quit":
        print("Bot: Goodbye!")
        break

    # Check knowledge base first
    answer = get_intent_response(user)

    if answer:
        print("Bot:", answer)
        continue

    # Otherwise ask the AI
    messages.append({
        "role": "user",
        "content": user
    })

    response = ollama.chat(
        model="llama3.2",
        messages=messages
    )

    bot = response["message"]["content"]

    print("Bot:", bot)

    messages.append({
        "role": "assistant",
        "content": bot
    })
```

```python
import sqlite3
import datetime
```

```python
conn = sqlite3.connect("chatbot.db")

cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS chat_history (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    sender TEXT,
    message TEXT,
    timestamp TEXT
)
""")

conn.commit()

print("Database created successfully!")
```

```python
def save_message(sender, message):

    timestamp = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")

    cursor.execute("""
    INSERT INTO chat_history(sender, message, timestamp)
    VALUES (?, ?, ?)
    """, (sender, message, timestamp))

    conn.commit()
```

```python
save_message("User", "Hello")

save_message("Bot", "Hi! How can I help?")
```

```python
cursor.execute("SELECT * FROM chat_history")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
memory = {}

def remember_name(user_input):

    if "my name is" in user_input.lower():

        name = user_input.split("is")[-1].strip()

        memory["name"] = name

        return f"Nice to meet you, {name}!"

    if "what is my name" in user_input.lower():

        if "name" in memory:

            return f"Your name is {memory['name']}."

        else:

            return "I don't know your name yet."

    return None
```

```python
print(remember_name("My name is Saksham"))

print(remember_name("What is my name"))
```

```python
import ollama

messages = []

print("=" * 60)
print("🤖 Smart AI Chatbot")
print("Type 'quit' to exit")
print("=" * 60)

while True:

    user = input("You: ")

    if user.lower() == "quit":
        break

    save_message("User", user)

    # Check memory
    reply = remember_name(user)

    if reply:
        print("Bot:", reply)
        save_message("Bot", reply)
        continue

    # Check intents
    intent_reply = get_intent_response(user)

    if intent_reply:
        print("Bot:", intent_reply)
        save_message("Bot", intent_reply)
        continue

    # AI response
    messages.append({
        "role": "user",
        "content": user
    })

    response = ollama.chat(
        model="llama3.2",
        messages=messages
    )

    bot = response["message"]["content"]

    print("Bot:", bot)

    save_message("Bot", bot)

    messages.append({
        "role": "assistant",
        "content": bot
    })

print("Goodbye!")
```
